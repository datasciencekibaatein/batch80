```python
import pandas as pd
```

```python
n = 100
df = pd.DataFrame({
    "customer_id":rng.integers(1000,9999,size=n),
    "salary":np.round(rng.uniform(25000,120000,size=n),2),
    "city":rng.choice(['delhi','mumbai','lucknow','banglore','pune'],size=n),
    "is_active":rng.choice([True,False],size=n),
    "join_date":pd.to_datetime("2023-01-01") + pd.to_timedelta(rng.integers(0,365,size=n),unit="D"),
    "department":rng.choice(['HR','IT','Finance','Sales'],size=n),
    "remarks":rng.choice(['good','averge','excellent','pending'],size=n),
    'rating':rng.choice([1,2,3,4,5,np.nan],size=n)
})
```
