<h2>What are missing values?</h2>

We encounter cells that are empty, null or undefined. These are called missing values - data point that were expected but were not recorded.

In python and pandas, missing values appear as NaN (not a number), None, or sometime as string like "N/A","-", or "unkown" that must be manually converted

<h2>Types of missing values</h2>

- MCAR - Missing completely at random
- MAR - Missing at random
- MNAR - Missing not at random

<h2>1. MCAR</h2>

The probability of a value being missed has absolutely no relationship with any variable in dataset.

<h2>2. MAR</h2>

The missingness is related to observed data, but not the missing values themselves.

<h2>3. MNAR</h2>

The missingness depends on the observed value itself. This is hardest to handle.

```python
import pandas as pd
import numpy as np

data = {
    'Name':["Rohan","Krish","Bob","Sohit","Kashish","Mayank"],
    'Age':[24,np.nan,34,23,np.nan,40],
    'Score':[85,90,np.nan,88,92,95],
    "Attendance":[100,90,80,np.nan,90,85]
}

df = pd.DataFrame(data)
```

```python
df.isnull().mean()*100
```

<h2>Row deletion</h2>

<h3>When it is good to use?</h3>

- Low percentage: if < 5% data is missing
- MCAR: The missingness has no hidden pattern
- Large Dataset: If you have 1 million rows losing 1000 wont hurt your model's performance.


<h3>When not to use it?</h3>

- Small dataset: if you only have 100 rows, every single rows is precious
- Bias risk: If only "High Income" individuals refused to answer the sruvey, dropping those rows remove "high income" representation entirely

```python
df_dropped = df.dropna()
```

```python
df_subset = df.dropna(subset=['Score','Attendance'])
```

<h2>Column Deletion</h2>

<h3>When it is good to Use</h3>

- High missingness: eg > 50,60 ,70...
- Irrelevance: If the feature is not crucial for the prediction

<h3>When it is not recommended</h3>

- Domain importance

```python
df['Notes'] = [np.nan,"Good",np.nan,np.nan,np.nan,"Average"]
```

```python
df
```

```python
missing_percentage = df.isnull().sum()/len(df)*100
```

```python
missing_percentage.index
```

```python
missing_percentage.values
```

```python
limit = 30

columns_to_drop = missing_percentage[missing_percentage>limit].index

columns_to_drop 
```

```python
df_reduced = df.drop(columns=columns_to_drop)
```
