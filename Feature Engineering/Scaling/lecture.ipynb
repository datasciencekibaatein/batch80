<h1>Scaling</h1>

Transforming features so they are on similar scale

```python
import pandas as pd
import numpy as np

data = {
    'age':np.random.randint(20,55,100),
    'salary':np.random.randint(30000,100000,100)
}

df = pd.DataFrame(data)
df.head(10)
```

<h2>Standard Scaler</h2>

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['age_scaled','salary_scaled']] = scaler.fit_transform(df[['age','salary']])
```

```python
df.head(10)
```

```python
df['age_scaled_manual'] = (df['age']-df['age'].mean())/df['age'].std()
```

```python
df['actual_age_manual'] = df['age_scaled_manual']*df['age'].std() + df['age'].mean()
```

```python
df[['age_scaled_inverse','salary_scaled_inverse']] = scaler.inverse_transform(df[['age_scaled','salary_scaled']] )
```

<h2>Min-Max Scaler</h2>

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['age_scaled','salary_scaled']] = scaler.fit_transform(df[['age','salary']])
```

```python
df.head(10)
```

```python
age_min = df['age'].min()
age_max = df['age'].max()
```

```python
df['age_scaled_manual'] = (df['age'] - age_min)/(age_max - age_min)
```

```python
df[['age_inverse','salary_inverse']] = scaler.inverse_transform(df[['age_scaled','salary_scaled']])
```
