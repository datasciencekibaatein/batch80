```python
import numpy as np
import pandas as pd


np.random.seed(42)

n = 100

df = pd.DataFrame(
    {
        "Age":np.random.randint(18,55,n),
        "Salary":np.random.randint(30000,100000,n),
        "experience":np.random.randint(1,30,n),
        "hours_worked":np.random.randint(4,12,n)
    }
)

df.head(10)
```

```python
df.head(20)
```

```python
df['Salary_per_exp'] = df['Salary']/df['experience']
```

```python
df['age_exp_interaction'] = df['Age']*df['experience']
```

```python
df['Age_group'] = pd.cut(
    df['Age'],
    bins=[18,28,42,60],
    labels = ['young','middle','senior']
)
```

```python
df['salary_hike'] = np.where(df['experience']>10,df['Salary']+df['Salary']*0.1,df['Salary'])
```

```python
df['Work_hr_group'] = pd.cut(
    df['hours_worked'],
    bins=[3,7,10,12],
    labels = ['least','middle','much']
)
```

```python
conditions = [df['hours_worked']<=7,df['hours_worked']<=10,df['hours_worked']<=12]

choices = ["A","B","C"]

df['ch'] = np.select(conditions,choices)
```
