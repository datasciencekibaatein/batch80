<h1><strong>Encoding Categorical Data</strong></h1>

```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 100

df = pd.DataFrame(
    {
        "age":np.random.randint(18,60,n),
        "salary":np.random.randint(30000,100000,n),
        "city":np.random.choice(["Delhi","Mumbai","Pune","Noida"],n),
        "gender":np.random.choice(["Male","Female"],n),
        "education":np.random.choice(['High School','Graduate','Post Graduate'],n)
    }
)
```

```python
df.head()
```

<h2>Label Encoder</h2>

Label encoder assign unique number to each category.

```python
pip install scikit-learn
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['City_label'] = le.fit_transform(df['city'])
```

```python
df.head(10)
```

<h2>One Hot Encoder</h2>

Gives you sparse matrix(0,1) and generate columns to each category.

```python
from sklearn.preprocessing import OneHotEncoder

oe = OneHotEncoder()

encoded = oe.fit_transform(df[['city']]).toarray()
```

```python
df[["Delhi","Mumbai","Noida","Pune"]] = encoded
```

```python
df.head(10)
```

<h2>Ordinal Encoder </h2>

Used when categories have order

```python
from sklearn.preprocessing import OrdinalEncoder

categories = [["High School","Graduate","Post Graduate"]]

oe = OrdinalEncoder(categories=categories)

df['education_encoded'] = oe.fit_transform(df[['education']])
```
