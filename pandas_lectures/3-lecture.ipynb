Data Integration (Merging all sheets)

```python
df = employees.merge(departments,on='DeptID',how='left')
```

```python
df = df.merge(locations,on="LocationID",how="left")
```

```python
df = df.merge(jobs,on='JobID',how='left')
df = df.merge(attendance,on='EmployeeID',how='left')
df = df.merge(attrition,on='EmployeeID',how='left')
```

```python
df.head()
```

```python
df['ExitReason'].notnull()
```

```python
df['Is_Attrited'] = df['ExitReason'].notnull()
```

```python
df.head()
```

```python
print(df.columns)
```

```python
df['Attrition_numeric'] = df['Is_Attrited'].astype(int)
```

Key HR Metrics

```python
metrics = {
    'Total Employees':len(df),
    'Average Salary':df['Salary'].mean(),
    'Attrition Rate %': (df['Is_Attrited'].sum()/len(df))*100,
    "Avg engagement Score":df['EngagementScore'].mean(),
    'Avg Satisfaction Score':df['SatisfactionScore'].mean(),
    'Total Bonus pool':df['Bonus'].sum(),
    'Avg Training Hours':df['TrainingHours'].mean(),
    'Avg Age':df['Age'].mean(),
    'Departments':df['Department'].nunique(),
    'Avg year since Promotion':df['YearsSincePromotion'].mean()
}
```

```python
for k,v in metrics.items():
    print(f"{k}: {v:,.2f}")
```
