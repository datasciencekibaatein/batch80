<h1><strong>Indexing</strong></h1>

- Label based
- Position based

```python
import pandas as pd
import numpy as np

```

```python
df = pd.DataFrame({
    "Name":["Amit","Riya","John","Sara","Amit","Sneha"],
    "Age":[22,25,30,None,20,31],
    "Salary":[50000,60000,None,65000,None,63000],
    "City":["Delhi",np.nan,"Delhi","Pune","Delhi",None]
},index=["a","b","c","d","e","f"])
```

```python
df
```

Label based indexing

```python
df.loc['a':'d',["Name","Age","Salary"]]
```

```python
df.loc['a'].values
```

Position based Indexing

```python
df.iloc[0:3,1:3]
```

```python
df['Age']>25
```

```python
df[(df['Age']>20) & (df['City']=="Delhi")]
```

```python
df.filter(items=['Name','Age'])
```

```python
df.filter(like='A')
```

```python
df.filter(items=['a',"c",'b','d'],axis=0)
```

```python
df[["Age","Salary"]].query("Salary > 50000 or Age > 30")
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
```

```python
df['City'] = df['City'].fillna("Lucknow")
```

```python
df.groupby("City")['Salary'].sum()
```

```python
df.groupby('City')['Salary'].agg(['sum','mean','min','max'])
```

```python
df['Job_Role'] = ['IT','IT','Finance','Sales','Sales','IT']
```

```python
df
```

```python
df.groupby(['Job_Role','City'])['Salary'].agg(['sum'])
```

```python
df.groupby('Job_Role').size()
```

```python
df.pivot_table(values="Salary",index="City",aggfunc="sum",columns="Job_Role")
```

```python
df['City'] = df['City'].replace("Delhi","New Delhi")
```
