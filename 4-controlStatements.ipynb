```python
# if statement
age = 17
print(age>=18)
if age>=18:
    print("you can vote")#indentation
```

```python
marks = int(input("Enter the marks: "))

if marks>=90:
    print("Grade A")
elif marks >=80:
    print("Grade B")
elif marks >=70:
    print("Grade C")
elif marks >=60:
    print("Grade D")
else:
    print("Fail")
```

```python
registered = "yes"
if registered =="yes":
    age = int(input("Enter the age: "))
    if age>=18:
        print("you can vote")
    else:
        print("Can't vote")
else:
    print("Registeration is recommended")
    
    
```
