WAP to find common element in two lists

```python
lst1 = [2,3,4,6,5]
lst2 = [3,9,8,2,8]

common = []

for i1 in lst1:
    for i2 in lst2:#[3,9,8,2,8]
        if i1 == i2:
            common.append(i1)

print(common)
```

WAP to count the occurence of an element in a tuple , without using inbuilt function

```python
tp = (12,23,23,10,45,90,65,12)

element = int(input("Enter the element to count occurence: "))
counter = 0
for i in tp:
    if i == element:
        counter+=1

print(f"Occurence of {element}: {counter} ")
```

WAP to check is a number armstrong or not

```python
num = int(input("Enter the number: "))
n = num
orginal_num = num
#count the digits in number
counter = 0
while num>0:
    num = num // 10
    counter = counter + 1


res = 0
while n>0:
    re = n % 10
    res = res + re**counter
    n = n //10

if res == orginal_num:
    print("Armstrong number")
else:
    print("Its not")
```
