<h2><strong>Encapsulation</strong></h2>

Encapsulation means wrapping data and methods into a single unit and restricting direct access

<strong>Why encapsulation is important ?</strong>

1. Data hiding
2. Security
3. Control over data
4. Easy maintenance
5. Flexibility

```python
class Bank:
    def __init__(self):
        self.balance = 1000

b = Bank()
b.balance = -5000

print(b.balance)
```

<strong>Encapsulation achieved using:</strong>

- public members
- protected members
- private members

<strong>Public</strong>

```python
class Student:
    def __init__(self):
        self.name = "Rahult" # this instance variable is public 

s = Student()
print(s.name)
```

<strong>Protected</strong>

```python
# Single underscore

class Student:
    def __init__(self):
        self._marks = 90  # this instance variable is protected

s = Student()
print(s._marks)
```

<strong>Private</strong>

```python
# double underscore

class Student:
    def __init__(self):
        self.__marks = 90


s = Student()
print(s.__marks)
```

<h2><strong>Inheritance</strong></h2>

Inheritance allows us to use parents class properties in child class.

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def eat(self):
        print("Eats")


class Dog(Animal):
    def speak(self):
        print("barks")


class Cat(Animal):
    def speak(self):
        print("meow")


a = Animal("jack")

c = Cat("cat name")
c.name


```

```python
class Bank:
    def __init__(self,initial_balance):
        self.__balance = initial_balance

    def credit(self,amount):
        if amount >0:
            self.__balance += amount
            print(f"Amount credited:{amount}")
        else:
            print("Invalid amount")

    def debit(self,amount):
        if amount>0:
            if amount<=self.__balance:
                self.__balance-=amount
                print(f"Amount debited:{amount}")
            else:
                print("Insufficient balance")
        else:
            print("Invalid amount")

    def get_balance(self):
        return self.__balance    

class User(Bank):
    def __init__(self,name,initial_balance):
        super().__init__(self,initial_balance)
        self.name = name

    def show_balance(self):
        print(f"{self.name}'s current balance:{self.get_balance()}")
```

```python
class Bank:
    def __init__(self,initial_balance):
        self.__balance = initial_balance

    def credit(self,amount):
        self.__balance += amount
       

    def debit(self,amount):
        self.__balance-=amount

    def get_balance(self):
        return self.__balance    

class User(Bank):
    def __init__(self,name,initial_balance):
        Bank.__init__(self,initial_balance)
        self.name = name

    def show_balance(self):
        print(f"{self.name}'s current balance:{self.get_balance()}")
```

```python

user1 = User(input("Enter the name: "),int(input("Enter the initial amount: ")))
user1.show_balance()
user1.credit(int(input("Credit amount: ")))
user1.show_balance()
user1.debit(int(input("Debit amount: ")))
user1.show_balance()
```

<strong>Types of inheritance</strong>

3. Multilevel

<strong>Multiple Inheritance</strong>

If one class is inherting more than one class

```python
class Father:
    def skills(self):
        print("worker")

class Mother:
    def skill(self):
        print("cooking")
    
class Child(Father,Mother):
    pass

c = Child()
c.skills()
```

Multilevel Inheritance

```python
class Grandparent:
    def home(self):
        print("Grandparent's home")

class Parent(Grandparent):
    def car(self):
        print("Parent's car")
    
class Child(Parent):
    def bike(self):
        print("Child's bike")

c = Child()
c.home()
c.car()
c.bike()
```

Hierarchical inheritance

```python
class Parent:
    def home(self):
        print("Home")

class Child1(Parent):
    pass

class Child2(Parent):
    pass


c1 = Child1()
c2 = Child2()
c1.home()
c2.home()
```

super()

```python
class Parent:
    def show(self):
        print("Parent")

class Child(Parent):
    def s(self):
        super().show()
        print("child")

c = Child()
c.s()
```

<strong>1. Function Polymorphism</strong>

<strong>2. Operator Overloading</strong>

```python
print(5+3)
print("hi"+"guest")
print([1,2]+[3,4])
```

```python
class Student:
    def __init__(self,marks):
        self.marks = marks

    def __add__(self,o):
        return self.marks + o.marks
    
s1 = Student(89)
s3 = Student(90)

print(s1+s3)

```

<strong>3. Method Overriding</strong>

```python
class Animal:
    def sound(self):
        print("Animal makes sound")
    
    def hi(self):
        print("i am inside animal")
    
class Dog(Animal):
    def sound(self):
        print("dog makes sound")
    
class Cat(Animal):
    def sound(self):
        print("Cat Meows")


a = Animal()
d = Dog()
c = Cat()

d.sound()
```

<strong>4. Duck Typing</strong>

```python
class Bird:
    def fly(self):
        print("Bird can fly")

class Airplane:
    def fly(self):
        print("Airplane can fly")

def make_it_fly(obj):
    obj.fly()


# b = Bird()
# b.fly()
# a = Airplane()
# a.fly()

make_it_fly(Airplane())
```
