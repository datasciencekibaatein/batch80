<h2>What is object ?</h2>

An object is an instance of a class.

```python
class Student1:
    pass


s1 = Student1() #object of class Student
print(type(s1))
print(s1.name)
```

```python
class Student2:
    def __init__(self,name,city,course): #constructor
        self.name = name
        self.city = city
        self.course = course


s = Student2("Rahul","Delhi","math")
print(s.city)
print(type(s))
```

<h2>Constructor</h2>

A constructor is automatically called when an object is created.

```python
class Vehicle:
    def __init__(self,name): #constructor
        self.name = name
        print("Constructor evoked",self.name)


v = Vehicle("hyundai")
print(v.name)
```

<h2>Instance Variables and methods</h2>

self refers to the current class/object

```python
class Student:
    def __init__(self,name,marks):
        self.m = marks #instance variable
        self.n = name #instance variable
        
    
    def detail(self):#instance method
        print("Name: ",self.n)
        print("Marks:",self.m)
    
    def acc(self):
        self.detail()
    
```

```python
student = Student("Krish","56")
student.acc()
```
