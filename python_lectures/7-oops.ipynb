A constructor is automatically called when an object is created.

<h2>Instance Variables and methods</h2>

self refers to the current class/object

```python
class Student:
    def __init__(self,name,marks):
        self.m = marks #instance variable
        self.n = name #instance variable
        
    
    def detail(self):#instance method
        print("Name: ",self.n)
        print("Marks:",self.m)
    
    def acc(self):
        self.detail()
    
```

```python
student = Student("Krish","56")
student.acc()
```

<h2><strong>Encapsulation</strong></h2>

Encapsulation means wrapping data and methods into a single unit and restricting direct access

<strong>Why encapsulation is important ?</strong>

1. Data hiding
2. Security
3. Control over data
4. Easy maintenance
5. Flexibility

```python
class Bank:
    def __init__(self):
        self.balance = 1000

b = Bank()
b.balance = -5000

print(b.balance)
```

<strong>Encapsulation achieved using:</strong>

- public members
- protected members
- private members

<strong>Public</strong>

```python
class Student:
    def __init__(self):
        self.name = "Rahult" # this instance variable is public 

s = Student()
print(s.name)
```

<strong>Protected</strong>

```python
# Single underscore

class Student:
    def __init__(self):
        self._marks = 90  # this instance variable is protected

s = Student()
print(s._marks)
```

<strong>Private</strong>

```python
# double underscore

class Student:
    def __init__(self):
        self.__marks = 90


s = Student()
print(s.__marks)
```

<h2>Access using method</h2>

```python
class Student:
    def __init__(self,name,age,marks):
        self.name = name#public instance variable
        self._age = age #protected instance variable
        self.__marks = marks #private instance variable

    def detail(self):
        print("Name: ",self.name)
        print("Age: ",self._age)
        print("Marks :",self.__marks)
```

```python
s = Student("Ankit",26,35)


s.detail()
```

```python
class Employee:
    def __init__(self,salary):
        self.__salary = salary

    def get_salary(self):
        print("inside getsalary: ",self.__salary)
        return self.__salary
    
    def set_salary(self,salary):
        if salary > 0:
            self.__salary = salary
```

```python
e = Employee(5000)

print(e.get_salary())


e.set_salary(10000)
print(e.get_salary())
```

<h2><strong>Inheritance</strong></h2>

Inheritance allows us to use parents class properties in child class.

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def eat(self):
        print("Eats")


class Dog(Animal):
    def speak(self):
        print("barks")


class Cat(Animal):
    def speak(self):
        print("meow")


a = Animal("jack")

c = Cat("cat name")
c.name


```

<strong>Types of inheritance</strong>

1. Single
2. Multiple 
3. Multilevel
4. Hierarchical
