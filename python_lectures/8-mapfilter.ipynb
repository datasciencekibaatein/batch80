Map

map() applies a function to every element of collection/iterable(list,tuple)

```python
nums = [1,2,3,4]

res = []
for i in nums:
    res.append(i**2)

res
```

```python
def square(x):
    return x**2
```

```python
result = list(map(square,nums))
```

```python
result
```

```python
list(map(lambda x:x**2,nums))
```

```python
print(type(input()))
```

```python
arr = ["1","2","3","5"]
```

```python
arr
```

```python
res = 0
for i in list(map(int,arr)):
    res = res + i
```

```python
res
```

Multiple Iterable

```python
a = [1,2,3,4]
b = [4,5,6,7]

result = list(map(lambda x,y:x+y,a,b))
```

```python
result
```

```python
names = ["rohit","krish","amit"]

result = list(map(str.upper,names))
```

```python
result
```

<h2>Filter</h2>

filter() selects elements from iterable based on condition(True/False)

```python
nums = [34,23,45,45,7,23,43,78,12]
```

```python
list(filter(lambda x:x%2==0,nums))
```

```python
data = ["hello","","python","","real"]
```

```python
res = list(filter(None,data))
```

```python
res
```

Square only even numbers

```python
lst = [4,5,3,2,8,10,16,71]
```

```python
list(map(lambda x:x**2,
    filter(lambda x:x%2==0,lst)))
```

Pythonic Alternative(List comprehension)

map alternative

```python
[x**2 for x in lst]
```

Filter Alternative

```python
[x for x in lst if x % 2==0]
```

```python
[x**2 for x in lst if x % 2 ==0]
```

```python
res = [x for x in lst]
```

```python
res
```

```python
res = []
for x in lst:
    if x%2==0:
        res.append(x**2)
```
