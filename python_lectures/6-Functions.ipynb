<h2><strong>Functions in python</strong></h2>

A functions is a block of reusable code that performs a specific tasks.

<strong>Why functions are important:</strong>

1. code reusability
2. better organization
3. reducing repetition
4. improving readability
5. Modular programming

<strong>Components</strong>

- def -> keyword to define the functions
- function_name -> name of function
- parameters -> inputs
- return -> output

```python
def greet(name): #function definition
    return "Hello , I am "+name#function body
```

```python
greet("ram") #function call
greet("rohit")
```

```python
def factorial(n):
    fact = 1
    for i in range(1,n+1):
        fact = fact * i
    
    return fact

```

```python
factorial(5)
```

Arbitrary Arguments (*args)

```python
def add_numbers(*numbers):
    total = 0
    print(*numbers)
    for num in numbers:
        total = total + num
    return total
```

```python
add_numbers(2,3,4,45,23,6,23,765,21,546,7,90)
```

<h3>Arbitrary keyword arguments(**)</h3>

Accept unlimited named arguments

```python
def student_details(**info):
    for key,value in info.items():
        print(key,":",value)


student_details(name="aman",age=25,course='ai')
```

<h3>Recursive  functions</h3>

A function calling itself is called recursion/recursive function.

```python
def greet():
    print("hello")
    greet()


greet()
```

Calculate the factorial using recursion?

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

```python
factorial(5)
```

<h3>Nested functions</h3>

```python
def outer():
    print("this is outer print")
    
    def inner():
        print("this is inner print")
    
    inner()
    print("print after inner function")
    inner()


outer()
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
```

```python

def check_strong_number(n):
    num = n
    res = 0 
    while n>0:
        re = n % 10
        fact = factorial(re)
        res = res + fact
        n = n //10
    print(f"{num} is a strong number ") if num==res else print(f"{num} is not strong number")
```

```python
check_strong_number(int(input("Enter the number to check : ")))
```
