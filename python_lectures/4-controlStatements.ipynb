```python
#2. WAP to check whether number is even or odd.

num = int(input("Enter the number: "))
if num%2==0:
    print(num," is even number.")
else:
    print(num," is odd number.")
```

```python
#3. WAP to find the largest of three number
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))
num3 = int(input("Enter num3: "))
if num1>=num2 and  num1>=num3:
    print(num1,"is greater")
elif num2>=num1 and num2>=num3:
    print(num2,"is greater")
else:
    print(num3,"is greater")
```

```python
#4. WAP to check whether a year is leap year or not.
year = int(input("Enter the year: "))
if (year%4==0 and year %100!=0) or (year % 400==0):
    print(year,"is leap year.")
else:
    print(year,"is not a leap year.")
```

```python
#5. WAP to check whether a character is a vowel or not.
ch = input("Enter a character: ")
if ch in "aeiouAEIOU":
    print(ch,"is vowel")
else:
    print(ch,"is consonant.")
```

```python
#6. WAP to calculate the grade of a student based on marks
marks = int(input("Enter the marks:"))
if marks>=90:
    print("Grade A")
elif marks>=75:
    print("Grade B")
elif marks>=50:
    print("Grade C")
else:
    print("Fail")
```

```python
#7. WAP to check whether a number is divisible by both 3 and 5
num = int(input("Enter the number: "))
if num%3==0 and num%5==0:
    print(num,"Divisible by both 3 & 5")
elif num%3==0:
    print(num,"Only divisible by 3")
elif num%5==0:
    print(num,"Only divisible by 5")
else:
    print(num,"divisible by none of them")
```

```python
#8. WAP to check whether a number is 2-digits , 3-digits
num = int(input("Enter the number: "))
if num<10:
    print(num,"is single digit.")
elif num<100:
    print(num,"is two digits.")
elif num<1000:
    print(num,"is three digits")
else:
    print("more than tree digits")
```

```python
#9. WAP to check calculate the electricity bill based on units consumed (define your own slab ranges)
units = int(input("Enter the units consumed: "))
if units<=100:
    bill = units*2
elif units<=300:
    bill = units*4
else:
    bill=units*6
    
print("Electricity bill is: ",bill)
```

```python
#10. WAP to check whether a given day number(1-7) corresponds to weekday or weekend.
day = int(input("Enter the day number (1-7): "))
if day in [1,2,3,4,5]:
    print("Weekday")
elif day in [6,7]:
    print("Weekend")
else:
    print("Invalid input")
```

```python
#11. WAP to calculate the discount based on purchase amount: define own criteria
amount = float(input("Enter the purchase amount: "))
if amount>5000:
    discount = amount * 0.2
elif amount >=2000:
    discount = amount * 0.15
else:
    discount = 0
    
print("Discount: ",discount)
print("Amount after discount: ",amount-discount)
```

```python
#12. WAP to accept a number and display whether it is divisible by 2,3 or 5

num = int(input("Enter the number:" ))
if num%2==0:
    print(num,"divisible by 2")
elif num%3==0:
    print(num,"divisible by 3")
elif num%5==0:
    print(num,"divisible by 5")
else:
    print("not divisible by any of them")
```

```python
#13. WAP to create a simple calculator using if-elif-else (operations: +,-,*,/,).
num1 = float(input("Enter the first number: "))
op = input("Enter the operator(+,-,x,/): ")
num2 = float(input("Enter the second number: "))


if op == "+":
    print("Result(+):", num1 + num2)
elif op == "-":
    print("Result(-):", num1 - num2)
elif op == "x":
    print("Result(x):", num1 * num2)
elif op == "/":
    if num2!=0:
        print("Result(/):", num1 / num2)
    else:
        print("Division by zero is not allowed")    
else:
    print("Invalied operation")    

    
```
