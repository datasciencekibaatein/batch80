<h3>Set</h3>

- Set is unordered datastructure
- Does not allow duplicate values
- To create set must use curly braces

```python
st = {45,34,78,45,33,34}
```

```python
st
```

<strong>Functions of Set</strong>

add()

```python
st.add(56) 
st.add(33)
st
```

update()

```python
st.update({56,78,89,10}) # i need to pass iterable -> collection
st
```

difference between add and update

- if you want to insert single value use add
- if it is collection of data use update

remove()

```python
st.remove(56) # if element exist , it will remove , otherwise throw error
```

```python
st
```

discard()

```python
st.discard(56) # if element exists it will remove, if not , it will not throw error
```

clear()

```python
st.clear()
```

```python
st
```

Set Operations :
- union()
- intersection()
- difference()

```python
a = {2,3,4,5}
b = {4,5,6,7,8}
```

```python
b
```

union() operation

```python
a.union(b)
```

intersection() operation

```python
a.intersection(b)
```

difference() operation

```python
b.difference(a)
```
