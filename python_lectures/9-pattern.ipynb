```python
for i in range(1,4):
    print("*"*i)
```

```python
n = 5
i = 0
while i<n : 
    j =0
    while j<n:
        if i>=j:
            print("(",i ,"," , j,")",end="")
            #print("*",end=" ")          
        j=j+1
    i = i+1
    print()

```

Inverted Right Angle

```python
n = 3
for i in range(n,0,-1):
    print("*"*i)
```

```python
n = 5
i = n
while i>=1: 
    j =1
    while j<=i:
        #print("(",i ,"," , j,")",end="")
        print("*",end=" ")
        j=j+1
    i = i-1
    print()
```

Pyramid

```python
n = 5
for i in range(1,n+1):
    print(" "*(n-i) + "*"*(2*i-1))
```

Number Triangle

```python
n = 5
for i in range(1,n+1):
    for j in range(1, i+1):
        print(j+i,end=" ")
    print()
```

Alphabet pattern

```python
n = 5
for i in range(n):
    for j in range(i+1):
        print(chr(65+i),end=" ")
    print()
```

Hollow Square

```python
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j ==0 or j ==n-1:
            # print("(",i ,"," , j,")",end="")
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```
