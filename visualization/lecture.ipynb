```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
n = 300

sales = pd.DataFrame({
    'Age':np.random.randint(18,60,n),
    'Salary':np.random.randint(25000,150000,n),
    'Experience':np.random.randint(1,20,n),
    'Department':np.random.choice(['IT','HR','Finance','Marketing'],n),
    'Gender':np.random.choice(['Male','Female'],n),
    'Performance_score':np.random.randint(50,100,n),
    'Sales':np.random.randint(5000,50000,n)
})
```

```python
sales.head()
```

Line Plot

```python
x = np.arange(1,11)
y = x * 2
plt.figure(figsize=(8,5))
plt.plot(x,y)
plt.title("Basic line plot")
plt.xlabel("X Values")
plt.ylabel("y values")
plt.show()
```

```python
x = np.arange(1,11)
y = x * 2
plt.figure(figsize=(8,5))
plt.plot(x,y,color='red',linestyle='--',marker='o',markersize=12)
plt.title("Basic line plot")
plt.xlabel("X Values")
plt.ylabel("y values")
plt.grid(True)
plt.show()
```

```python
x = np.arange(1,11)
y = x * 2
plt.figure(figsize=(8,5))
plt.plot(x,x*2,label='2x')
plt.plot(x,x*3,label='3x')
plt.plot(x,x*4,label='4x')
plt.title("Basic line plot")
plt.xlabel("X Values")
plt.ylabel("y values")
plt.legend()
plt.show()
```

Scatter plot

```python
plt.figure(figsize=(8,5))
plt.scatter(sales['Age'],sales['Salary'])
plt.title("Age Vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(sales['Age'],sales['Salary'],c = sales['Performance_score'],cmap='viridis')
plt.title("Age Vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.colorbar()
plt.show()
```

Bar Chart

```python
department_counts = sales['Department'].value_counts()

department_counts.index
```

```python
plt.figure(figsize=(8,5))
plt.bar(department_counts.index,department_counts.values)
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.barh(department_counts.index,department_counts.values)
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

```python
male = [20,35,30]
female = [25,32,34]

labels = ['IT','HR',"Finance"]

plt.figure(figsize=(8,5))
plt.bar(labels,male,label='Male')
plt.bar(labels,female,bottom=male,label='female')
plt.legend()
plt.title("stacked bar chart")
plt.show()
```

```python
female = [25,32,34]
male = [20,35,30]
labels = ['IT','HR',"Finance"]

plt.figure(figsize=(8,5))
plt.bar(labels,male,label='Male')
plt.bar(labels,female,bottom=male,label='female')
plt.legend()
plt.title("stacked bar chart")
plt.show()
```

Histogram Chart

```python
plt.figure(figsize=(8,5))
plt.hist(sales['Age'],bins=20,edgecolor='black',color='orange')
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequencey")
plt.show()
```

Pie Chart

```python
labels = department_counts.index
sizes = department_counts.values

plt.figure(figsize=(8,8))
plt.pie(sizes,labels=labels,autopct='%1.2f%%')
plt.title("Department Percentage")
plt.show()
```

Box Plot

```python
plt.figure(figsize=(8,5))
plt.boxplot(sales['Salary'])
plt.title("Salary box plot")
plt.show()
```
