```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
n = 300

sales = pd.DataFrame({
    'Age':np.random.randint(18,60,n),
    'Salary':np.random.randint(25000,150000,n),
    'Experience':np.random.randint(1,20,n),
    'Department':np.random.choice(['IT','HR','Finance','Marketing'],n),
    'Gender':np.random.choice(['Male','Female'],n),
    'Performance_score':np.random.randint(50,100,n),
    'Sales':np.random.randint(5000,50000,n)
})
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(sales['Age'],fill=True)
plt.title("KDE Plot")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=sales['Salary'],y=sales['Department'])
plt.title("Box plot of statistics")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(x=sales['Department'],y=sales['Salary'])
plt.title("Violin plot")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Gender',data=sales)
plt.title("Gender count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.stripplot(x='Gender',y='Salary',data=sales)
plt.title("Strip plot")
plt.show()
```

```python
sns.jointplot(
    x='Age',
    y='Salary',
    data=sales,
    kind='scatter'
)
```

```python
fig, axes = plt.subplots(2,2,figsize=(12,8))

axes[0,0].hist(sales['Salary'])
axes[0,0].set_title('Histogram')

axes[0,1].boxplot(sales['Salary'])
axes[0,1].set_title("Box plot")

axes[1,0].scatter(sales['Age'],sales['Salary'])
axes[1,0].set_title("Scatter plot")

axes[1,1].bar(sales['Department'],sales['Salary'])
axes[1,1].set_title("bar chart")

plt.tight_layout()
plt.savefig("charts.jpg")
plt.show()
```
